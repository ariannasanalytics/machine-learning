# Building the Main Algorithm

Now, as we build this, we will go through the program chunk by chunk and explain each line's purpose. 

```python
    while current_line:
        next_line = file.readline()
        stripped_line = current_line.strip()
```

This code chunk is establishing the variable <span style= color:#FF6FB5>current_line</span>, and also telling the program where to start reading in the file (the first line).

The <span style= color:#FF6FB5>while current_line</span> statement holds open the current line as it is being processed. Next we define the lookahead variable <span style= color:#FF6FB5>next_line</span> to establish our conditions with our CHAPTER I start point and <span style= color:#FF6FB5>stripped_line</span> to make sure all whitespace bookending and lines are gone.

The part below is important to ensure at the end of the text, or THE END, the extra things written after the final chapter concludes are not put into a specific chapter. Once the end of the story is found, the final chapter will be wrapped up and no more lines will be parsed and stored away. This also properly closes the final chapter. 

```python
        if stripped_line.upper() == "THE END":
            current_chapter.append(current_line.strip() + "\n")
            print(f"Found 'THE END' at line: {stripped_line}. Stopping processing.")
            break
```

This can be useful when separating any kind of document, since most books and articles have some amount of introductory information at the top we may not need an individual file for. For example, if we wanted a different file for each section of a research paper, there would likely be a lot of extra information about publishing and licensing before the abstract and after the conclusion. 

Next, the <span style= color:#FF6FB5> stripped_next</span> variable: this prepares the _next_ _line_ of text by removing any of the whitespace bookending. If there is no next line, the value of None is assigned, which prevents processing a next line that does not exist. This helps differentiate empty lines (which still exist) from reaching the true end of the file. The actual end where no further lines exist in the file is marked with "none". 

```python
       stripped_next = next_line.strip() if next_line else None
```

Ok so now we understand what the program will do when it finds the end of the story, and how it moves through the lines of the file. Now let's look to the beginning again. Since our aim is to divide the file into separate files for the chapter, how will our program know to omit the extra information that comes before the first chapter? Let's investigate.


This first line makes sure no automatic processing occurs over the whole text until the correct <span style=color:#2277DD;>CHAPTER I </span>is found. We don't need to save any text to any of our lists until we know for sure it is part of the first chapter. 

Additionally, in doing so, we are able to avoid giving the table of contents their own files since it contains the words <span style=color:#2277DD;>"CHAPTER I"</span> and that of every other chapter, which would trigger a false positive identification of the actual start for every single one. 

In order for this to happen successfully, the next line confirms that there is not another word <span style=color:#2277DD;>"Chapter"</span> aka the chapter marker at the beginning, ensuring it is the true start of the first chapter.

Well now it knows it has found the real start of the first chapter and will start compiling. So the variable <span style= color:#FF6FB5>chapter_found</span> is set to True.

```python
       if not chapter_found and stripped_line.startswith("CHAPTER I"):
            if not stripped_next.startswith(chapter_marker):  # table of contents safeguard
                chapter_found = True
```

Now that we establish what to do when the correct start is found, we must establish what to do if it is not. <span style= color:#FF6FB5>chapter_found</span> will always start as false since no valid chapter start has been found. Therefore, <span style= color:#FF6FB5>not chapter_found</span> means "if we haven't found a valid chapter yet"

<span style= color:#FF6FB5>current_line = next_line</span> means exactly what it sounds like: we advance a line forward, then we continue through the logic.

This block will not be executed once the correct starting point is found; we will then move on to processing the chapter.

```python
        if not chapter_found:
            current_line = next_line
            continue
```

Now we begin parsing through the text to find all of the rest of our chapters too. Remember, we chose this line-by-line method for several stylistic and functional reasons:

- This one line at a time method makes it easy to spot mistakes in code right away

Moving along to our next line of code:

We will begin parsing through the rest of the file to find all chapter markers. This line is making sure we confirm our <span style= color:#FF6FB5;>current stripped_line</span> starts with the chapter marker. If it does not, then we have ended up in a random no man's land in the middle of the text. That is not doing the thing we want. If it does though, we will be able to find where the error happened very quickly at least! In summary, this will help us separate chapters based on the marker.

```python
        if stripped_line.startswith(chapter_marker):
```

If we don't include this next code block, then we are not confirming that there is content to save in the current chapter. We could also not rule out the potential for existing in TOC since we would not be confirming the next line is free from the word chapter.

Our second line here saves the current chapter to its own nested list within <span style= color:#FF6FB5;>chapters</span> and the following one clears the working memory aka current_chapter so we can begin the next new chapter.

Now that we have arrived to the final chapter, we must make sure we save everything correctly. 

Say we omitted this from the code; it is possible the last chapter would not be saved since our existing loop would not automatically save it.

We have to be sure to append the last chapter if it exists, which it should based on how we would expect this program to function.

```python
if current_chapter:
    chapters.append(current_chapter[:])
```

This final chunk of code is to complete the final step of the task. We will give each chapter its own file and a name with the chapter number corresponding to the chapter number written in roman numerals.


We do NOT want to manually make a file for each chapter, or manually rename the roman numeral. While it is doable in this case, it is bad practice going forward if one were to deal with much larger data sets. 

To do this, we will use an enumerate function

Come important definitions: idx means index, so this first line is stating it will loop through each chapter within chapters starting from the first, giving them unique file names starting with number 1. This code using indexes allows for zeros in front of numbers in the beginning (as per normal index structure). 

Pay close attention to the open statement; 'w' is used instead of 'r' as is above, since in this case, we need to _write_ all of the text and information into the files.

```python
for idx, chapter in enumerate(chapters, start=1):
    chapter_filename = f"Dracula-Chapter-{idx:02}.txt"


    with open(chapter_filename, 'w', encoding='utf-8') as chapter_file:
        chapter_file.writelines(chapter)
```

If all has gone according to plan, this number should always say 27. Now let's put it all together! Once we run this code, all of the chapter files should show up in our Jupyter files saved. 

```python
chapters = []
current_chapter = []
chapter_marker = "CHAPTER"
chapter_found = False

with open('C:/Users/arial/PycharmProjects/pythonProject1/Midterm/pg345.txt', 'r', encoding='utf-8') as file:
    current_line = file.readline()
    while current_line:
        next_line = file.readline()
        stripped_line = current_line.strip()
        if stripped_line.upper() == "THE END":
            current_chapter.append(current_line.strip() + "\n")
            print(f"Found 'THE END' at line: {stripped_line}. Stopping processing.")
            break
        stripped_next = next_line.strip() if next_line else None
        if not chapter_found and stripped_line.startswith("CHAPTER I"):
            if not stripped_next.startswith(chapter_marker):  # table of contents safeguard
                chapter_found = True    
        if not chapter_found:
            current_line = next_line
            continue
        if stripped_line.startswith(chapter_marker):
            if current_chapter and not stripped_next.startswith(chapter_marker):
                chapters.append(current_chapter[:])  # save
                current_chapter = []  # clear
            current_chapter.append(current_line)
        else:
            current_chapter.append(current_line)
        current_line = next_line
if current_chapter:
    chapters.append(current_chapter[:])
for idx, chapter in enumerate(chapters, start=1):
    chapter_filename = f"Dracula-Chapter-{idx:02}.txt"


    with open(chapter_filename, 'w', encoding='utf-8') as chapter_file:
        chapter_file.writelines(chapter)
print(f"Successfully processed {len(chapters)} chapters.")
```
